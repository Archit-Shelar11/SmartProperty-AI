```python
def convert_price(price):

    if pd.isna(price):
        return np.nan

    price = str(price)

    if "Crore" in price:
        return float(price.split()[0]) * 10000000

    elif "Lac" in price:
        return float(price.split()[0]) * 100000

    else:
        return np.nan
```

```python
df['price_numeric'] = df['price'].apply(convert_price)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

df['price_numeric'] = imputer.fit_transform(df[['price_numeric']])
```

```python
df.sample(5)
```

```python
def convert_to_crore(price):

    if pd.isna(price):
        return np.nan

    price = str(price)

    if "Crore" in price:
        return round(float(price.split()[0]),2)

    elif "Lac" in price:
        return round(float(price.split()[0]) / 100,2)

    else:
        return np.nan
```

```python
df['price_crore'] = df['price'].apply(convert_to_crore)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

df['price_crore'] = imputer.fit_transform(df[['price_crore']])
```
